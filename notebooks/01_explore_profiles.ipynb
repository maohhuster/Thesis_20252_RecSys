# 01 — Explore Movie Profiles

Browse LLM-generated profiles, mood vectors, and themes.

**Prerequisites:** Download data with `make download-data` or have the profiles at `data/profiles/movie_profiles.json`.

```python
import json
import numpy as np
import pandas as pd
from pathlib import Path

# Load profiles
PROFILES_PATH = Path("../data/profiles/movie_profiles.json")
if not PROFILES_PATH.exists():
    PROFILES_PATH = Path("../code/profile_generator/llm-movie-profiler-v1-20260402/output/movie_profiles.json")

with open(PROFILES_PATH) as f:
    profiles = json.load(f)

print(f"Loaded {len(profiles)} profiles")
print(f"\nSample profile:")
sample = profiles[0] if isinstance(profiles, list) else list(profiles.values())[0]
for k, v in sample.items():
    if k == 'profile_text':
        print(f"  {k}: {v[:100]}...")
    else:
        print(f"  {k}: {v}")
```

## Mood Vector Distribution

Visualize the 10-axis mood vectors across all movies.

```python
import matplotlib.pyplot as plt
import seaborn as sns

MOOD_AXES = [
    "dark/light", "serious/playful", "slow/fast",
    "cerebral/visceral", "bleak/hopeful", "grounded/fantastical",
    "tense/relaxed", "conventional/experimental",
    "intimate/epic", "nostalgic/contemporary"
]

moods = np.array([p['mood_vector'] for p in (profiles if isinstance(profiles, list) else profiles.values())])

fig, ax = plt.subplots(figsize=(12, 5))
sns.boxplot(data=pd.DataFrame(moods, columns=MOOD_AXES), ax=ax)
ax.set_title("Mood Vector Distribution Across 10,381 Movies")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Theme Frequency

Most common themes across all movie profiles.

```python
from collections import Counter

all_themes = []
for p in (profiles if isinstance(profiles, list) else profiles.values()):
    all_themes.extend(p.get('key_themes', []))

theme_counts = Counter(all_themes).most_common(30)

fig, ax = plt.subplots(figsize=(12, 6))
themes, counts = zip(*theme_counts)
ax.barh(range(len(themes)), counts)
ax.set_yticks(range(len(themes)))
ax.set_yticklabels(themes)
ax.set_xlabel("Count")
ax.set_title("Top 30 Most Common Themes")
ax.invert_yaxis()
plt.tight_layout()
plt.show()
```
